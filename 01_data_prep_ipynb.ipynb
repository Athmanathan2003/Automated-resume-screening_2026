```python
# ========== CELL 6: LOAD DATASET + COMPUTE AVERAGE ==========
import pandas as pd
import numpy as np
import os

RAW_DIR = '/content/drive/MyDrive/ResumeProject/raw'
DATA_DIR = '/content/drive/MyDrive/ResumeProject/data'
os.makedirs(DATA_DIR, exist_ok=True)

# Load
df = pd.read_excel(f'{RAW_DIR}/cc.xlsx', sheet_name='Sheet1')

# Rename columns for consistency
df = df.rename(columns={
    'Skills and Achievements': 'Skills',
    'Job_type': 'Job_type',
    'Annotator-1': 'Ann1',
    'Annotator-2': 'Ann2',
})

# Compute average (paper's primary target)
df['Avg'] = (df['Ann1'] + df['Ann2']) / 2

print(f"✅ Loaded: {df.shape}")
print(f"\n📋 Columns: {list(df.columns)}")
print(f"\n📊 Domain distribution:")
print(df['Domain'].value_counts())
print(f"\n📊 Seniority distribution:")
print(df['Job_type'].value_counts())
print(f"\n📊 Target stats (Avg):")
print(df['Avg'].describe())
```

```python
# ========== CELL 7: TEXT CLEANING + COMBINING ==========
import re

def clean_text(text):
    """Basic cleaning: handle NaN, remove extra whitespace."""
    if pd.isna(text):
        return ""
    text = str(text)
    text = re.sub(r'\s+', ' ', text)   # collapse whitespace
    text = text.strip()
    return text

# Clean each section
df['Education']  = df['Education'].apply(clean_text)
df['Skills']     = df['Skills'].apply(clean_text)
df['Experience'] = df['Experience'].apply(clean_text)

# Combined text (for single-agent / TF-IDF / DNN / Transformers)
df['Combined'] = (
    "EDUCATION: " + df['Education'] +
    " SKILLS: " + df['Skills'] +
    " EXPERIENCE: " + df['Experience']
)

# Stats
print(f"✅ Text cleaned")
print(f"\n📏 Text length stats (characters):")
print(f"   Education:  min={df['Education'].str.len().min()},  max={df['Education'].str.len().max()},  mean={df['Education'].str.len().mean():.0f}")
print(f"   Skills:     min={df['Skills'].str.len().min()},  max={df['Skills'].str.len().max()},  mean={df['Skills'].str.len().mean():.0f}")
print(f"   Experience: min={df['Experience'].str.len().min()},  max={df['Experience'].str.len().max()},  mean={df['Experience'].str.len().mean():.0f}")
print(f"   Combined:   min={df['Combined'].str.len().min()},  max={df['Combined'].str.len().max()},  mean={df['Combined'].str.len().mean():.0f}")

# Show sample
print(f"\n👀 Sample combined text (first 300 chars of row 0):")
print(df['Combined'].iloc[0][:300] + "...")
```

```python
# ========== CELL 8 (FIXED): NORMALIZE JOB TYPE + STRATIFIED SPLIT ==========
from sklearn.model_selection import train_test_split
import pandas as pd

# ---- Step 1: Normalize Job_type (fix case inconsistency) ----
print("🔧 Normalizing Job_type...")
print(f"Before unique values: {df['Job_type'].unique()}")

df['Job_type'] = df['Job_type'].str.strip().str.title()   # normalize
print(f"After  unique values: {df['Job_type'].unique()}")

# Normalize Domain too (research assistant vs Research Assistant)
df['Domain'] = df['Domain'].str.strip().str.title()
print(f"Domain values: {df['Domain'].unique()}")

# ---- Step 2: Rebuild Strat_Key ----
df['Strat_Key'] = df['Domain'].astype(str) + "_" + df['Job_type'].astype(str)

print(f"\n📊 Normalized group counts:")
print(df['Strat_Key'].value_counts())
print(f"\nTotal groups: {df['Strat_Key'].nunique()}")

# ---- Step 3: Stratify by DOMAIN only (safe — all groups have >=50) ----
print(f"\n📊 Domain counts:")
print(df['Domain'].value_counts())

train_df, test_df = train_test_split(
    df,
    test_size=0.20,                 # 20% test
    stratify=df['Domain'],          # ✅ stratify by Domain only (safe)
    random_state=42
)

print(f"\n✅ Split complete:")
print(f"   Train: {len(train_df)} rows  ({len(train_df)/len(df)*100:.1f}%)")
print(f"   Test:  {len(test_df)} rows  ({len(test_df)/len(df)*100:.1f}%)")

print(f"\n📊 Train Domain distribution:")
print(train_df['Domain'].value_counts())
print(f"\n📊 Test Domain distribution:")
print(test_df['Domain'].value_counts())

print(f"\n📊 Train Job_type distribution:")
print(train_df['Job_type'].value_counts())
print(f"\n📊 Test Job_type distribution:")
print(test_df['Job_type'].value_counts())

print(f"\n📊 Target stats:")
print(f"   Train Avg: mean={train_df['Avg'].mean():.3f}, std={train_df['Avg'].std():.3f}, "
      f"min={train_df['Avg'].min():.3f}, max={train_df['Avg'].max():.3f}")
print(f"   Test  Avg: mean={test_df['Avg'].mean():.3f}, std={test_df['Avg'].std():.3f}, "
      f"min={test_df['Avg'].min():.3f}, max={test_df['Avg'].max():.3f}")
```

```python
# ========== CELL 9: WORDNET AUGMENTATION ==========
import random
from nltk.corpus import wordnet
import nltk

def get_synonyms(word):
    synonyms = set()
    for syn in wordnet.synsets(word):
        for lemma in syn.lemmas():
            syn_word = lemma.name().replace('_', ' ')
            if syn_word.lower() != word.lower() and ' ' not in syn_word:
                synonyms.add(syn_word)
    return list(synonyms)

def augment_text(text, replace_ratio=0.30, seed=None):
    if seed is not None:
        random.seed(seed)

    tokens = text.split()
    if len(tokens) == 0:
        return text

    eligible_idx = [i for i, t in enumerate(tokens)
                    if t.isalpha() and len(t) >= 4]

    if len(eligible_idx) == 0:
        return text

    n_replace = max(1, int(len(eligible_idx) * replace_ratio))
    replace_idx = random.sample(eligible_idx, min(n_replace, len(eligible_idx)))

    new_tokens = tokens.copy()
    for idx in replace_idx:
        word = tokens[idx]
        syns = get_synonyms(word)
        if syns:
            new_tokens[idx] = random.choice(syns)

    return ' '.join(new_tokens)

# Test
sample = df['Combined'].iloc[0]
aug_sample = augment_text(sample, replace_ratio=0.30, seed=42)

print("✅ Augmentation function ready")
print(f"\n📝 ORIGINAL (first 400 chars):")
print(sample[:400])
print(f"\n📝 AUGMENTED (first 400 chars):")
print(aug_sample[:400])
print(f"\n📏 Original length: {len(sample)}, Augmented length: {len(aug_sample)}")
```

```python
# ========== CELL 10: CREATE AUGMENTED SET + SAVE ==========
from tqdm import tqdm

print("⏳ Creating augmented version of training set...")
train_aug = train_df.copy()

aug_texts = []
for i, text in tqdm(enumerate(train_df['Combined'].tolist()), total=len(train_df)):
    aug = augment_text(text, replace_ratio=0.30, seed=i)
    aug_texts.append(aug)

train_aug['Combined'] = aug_texts
train_aug['is_augmented'] = 1

train_df_copy = train_df.copy()
train_df_copy['is_augmented'] = 0

# Combine: original + augmented
train_combined = pd.concat([train_df_copy, train_aug], ignore_index=True)

print(f"\n✅ Augmented set created:")
print(f"   Original train: {len(train_df_copy)}")
print(f"   Augmented train: {len(train_aug)}")
print(f"   Total train (combined): {len(train_combined)}")
print(f"   Test: {len(test_df)}")

# Save to Drive
train_df_copy.to_csv(f'{DATA_DIR}/train_orig.csv', index=False)
train_combined.to_csv(f'{DATA_DIR}/train_aug.csv', index=False)
test_df.to_csv(f'{DATA_DIR}/test.csv', index=False)

print(f"\n💾 Saved to Drive:")
print(f"   {DATA_DIR}/train_orig.csv  ({len(train_df_copy)} rows)")
print(f"   {DATA_DIR}/train_aug.csv   ({len(train_combined)} rows)")
print(f"   {DATA_DIR}/test.csv        ({len(test_df)} rows)")
```
